# **정규식**

정규식 공부

1. [] : 문자 클래스를 만드는 메타 문자인 [ ] 사이에는 어떤 문자도 들어갈 수 있다.

  즉, 정규 표현식이 [abc]라면 이 표현식의 의미는 ‘a, b, c 중 한 개의 문자와 매치’를 뜻한다.

  [] 안의 두 문자 사이에 하이픈(-)을 사용하면 두 문자 사이의 범위를 의미한다. 예를 들어 [a-c]라는 정규 표현식은 [abc]와 동일하고 [0-5]는 [012345]

  [a-zA-Z] : 모든 알파벳

  [0-9] : 모든 숫자

2. .(dot) 문자 - \n을 제외한 모든 문자

  a.b 는 "a + 모든_문자 + b" : a와 b라는 문자 사이에 어떤 문자가 들어가도 모두 매치

3. \* 문자

  *은 * 바로 앞에 있는 문자가 0부터 무한대까지 반복될 수 있다는 의미

4. \+ 문자

  +는 최소 1번 이상 반복될 때 사용한다. 즉, *가 반복 횟수가 0부터라면 +는 반복 횟수가 1부터인 것이다.

5. {m, n} 정규식을 사용하면 반복 횟수가 m부터 n까지인 문자와 매치할 수 있다. m 또는 n을 생략할 수도 있다.

  만약 {3,}처럼 사용하면 반복 횟수가 3 이상인 경우이고 {, 3}처럼 사용하면 반복 횟수가 3 이하인 경우를 의미한다. 생략된 m은 0과 동일하며, 생략된 n은 무한대의 의미를 갖는다.

  {1,}은 +, {0,}은 *와 동일하다

  ca{2,5}t : "c + a를_2~5회_반복 + t"

6. ? : ? 메타 문자가 의미하는 것은 {0, 1}이다.

  ab?c : "a + b가_있어도_되고_없어도_됨 + c"

7. 그 외 문자 클래스

  \d - 숫자와 매치된다. [0-9]와 동일한 표현식이다.

  \D - 숫자가 아닌 것과 매치된다. [^0-9]와 동일한 표현식이다.

  \s - 화이트스페이스(whitespace) 문자와 매치된다. [ \t\n\r\f\v]와 동일한 표현식이다. 맨 앞의 빈칸은 공백 문자(space)를 의미한다.

  \S - 화이트스페이스 문자가 아닌 것과 매치된다. [^ \t\n\r\f\v]와 동일한 표현식이다.

  \w - 문자+숫자(alphanumeric)와 매치된다. [a-zA-Z0-9_]와 동일한 표현식이다.

  \W - 문자+숫자(alphanumeric)가 아닌 문자와 매치된다. [^a-zA-Z0-9_]와 동일한 표현식이다.

8. 정규식을 이용한 문자열 검색

  match()	문자열의 처음부터 정규식과 매치되는지 조사한다.

  search()	문자열 전체를 검색하여 정규식과 매치되는지 조사한다.

  findall()	정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴한다.

  finditer()	정규식과 매치되는 모든 문자열(substring)을 반복 가능한 객체로 리턴한다.

compile을 사용하면 여러번 사용할 경우 일반 사용보다 더 빠른 속도를 보임
compile을 통해 정규표현식을 사용할 경우 re가 아닌 컴파일한 객체이름을 사용해야 함

```python
import time

normal_s_time = time.time()
r = 'ab.'
for i in range(1000) :
  re.match(r, 'abc')

print('일반 사용시 소요시간 :', time.time() - normal_s_time)

check = 'abc'
compile_s_time = time.time()
r = re.compile('ab.')

for i in range(1000) :
  r.match(check)

print('compile 사용시 소요시간 :', time.time() - compile_s_time)
```

findall() - 컴파일한 정규표현식을 이용해 정규 표현식과 맞는 모든 문자(열)을 리스트로 반환

re.findall('패턴 문자열', '문자열')

```python
import re
r = '[a-z]+'

print(re.findall(r,'aaa bbb ccc'))
print(re.findall(r,'aaa 000 bbb ***'))
print(re.findall(r,'000 aa0 bb! ___'))
```

```python
print(re.findall(r'[\d]','1ab 2cd e3f'))
print(re.findall(r'[\D]','w2%^ 3()'))
print(re.findall(r'[\D]+','w2%^ 3()'))
```

finditer()  - 컴파일한 정규 표현식을 이용해 정규 표현식과 맞는 모든 문자(열)을 iterator 객체로 반환


```python
iter1 = re.finditer(r'[\d]','1ab 2cd e3f')
iter2 = re.finditer(r'[\D]+','w2%^ 3()')
#print(iter1)

for c in iter1 :
  print(c)

print('-'*100)
for c in iter2 :
  print(c)
```

group() 메서드를 호출하면 매칭된 전체 문자열을 반환

```python
iter1 = re.finditer(r'[\d]','1ab 2cd e3f')
iter2 = re.finditer(r'[\D]+','w2%^ 3()')
#print(iter1)

for c in iter1 :
  print(c.group())

print('-'*100)
for c in iter2 :
  print(c.group())
```

한국어 정규식

match(패턴, string) : 컴파일한 정규표현식을 이용해 문자열이 정규 표현식에 맞는지 검사

```python
import re

check = '[ㄱ-ㅎ]+'

print(re.match(check, 'ㅎ 안녕하세요'))
print(re.match(check, 'ㅁㅅㅎ 안녕하세요 ㅎ'))
print(re.match(check, '가나다라'))
```

search(패턴, string) : match와 다르게 문자열의 전체를 검사

```python
check = '[ㄱ-ㅎ|ㅏ-ㅣ]+'   # OR 연산자

print(re.match(check, 'ㄱㅏ 안녕하세요'))
print(re.match(check, '안ㄱㅏ'))
print(re.search(check, '안ㄱㅏ'))
print(re.search(check, '안가'))
```

정규 표현식을 이용한 토큰화

단어 토큰화를 수행하는 RegexpTokenizer를 지원합니다. RegexpTokenizer()에서 괄호 안에 하나의 토큰으로 규정하기를 원하는 정규 표현식을 넣어서 토큰화를 수행합니다.

gaps=False (기본값): 주어진 정규 표현식에 매치되는 부분을 토큰으로 반환합니다.

gaps=True: 주어진 정규 표현식을 기준으로 문자열을 분할하고, 매치되지 않는 부분을 토큰으로 반환합니다.

```python
from nltk.tokenize import RegexpTokenizer

sentence = 'Where there\'s a will, there\'s a way. Right?'
tokenizer1 = RegexpTokenizer(r'[\w]+')
tokens1 = tokenizer1.tokenize(sentence)
print(tokens1)
```

```python
tokenizer2 = RegexpTokenizer(r'\s+', gaps=True)
tokens2 = tokenizer2.tokenize(sentence)
print(tokens2)
```

```python
tokenizer3 = RegexpTokenizer(r'\s+')
tokens3 = tokenizer3.tokenize(sentence)
print(tokens3)
```

# **문장 분석**

문장 분석 (Sentence Parsing, Noun Phrase Extraction, POS Tagging, Sentiment Analysis)

TextBlob의 Polarity (극성):

정의: Polarity는 텍스트의 감정이 얼마나 긍정적이거나 부정적인지를 나타냅니다.

범위: 값은 -1.0에서 1.0까지의 범위를 가지며,
1.0에 가까울수록 매우 긍정적인 감정을 의미하고,
-1.0에 가까울수록 매우 부정적인 감정을 의미합니다.

0은 중립적인 감정을 나타냅니다.

TextBlob의 Subjectivity (주관성):

정의: Subjectivity는 텍스트가 얼마나 주관적인지 또는 의견, 감정, 신념 등을 포함하고 있는지를 나타냅니다.

범위: 값은 0.0에서 1.0까지의 범위를 가지며,
1.0에 가까울수록 매우 주관적인 텍스트를 의미합니다.

0.0에 가까울수록 매우 객관적이고 사실에 기반한 텍스트를 의미합니다.

PoS(Part of Speech)태킹 : PoS는 품사를 의미. PoS 태깅은 각 품사를 태킹

개체명 인식(Named Entity Recognition)

nltk.download('maxent_ne_chunker')는 텍스트에서 명명된 개체(named entities)를 인식하고 추출하는 데 사용됩니다. 명명된 개체는 사람 이름, 조직 이름, 장소, 시간, 날짜, 금액 등과 같은 특정한 이름이 있는 항목을 의미합니다.

nltk.download('words')는 NLTK 라이브러리에서 제공하는 영어 단어 목록을 다운로드합니다. 이 데이터는 단어 목록, 즉 영어 단어의 집합을 포함하며, 이는 다양한 자연어 처리(NLP) 작업에서 유용하게 사용될 수 있습니다.

nltk.download('averaged_perceptron_tagger')는 텍스트의 각 단어에 대해 품사(예: 명사, 동사, 형용사 등)를 태깅하는 데 사용됩니다.

Words Corpus: 단순히 영어 단어들의 리스트로 구성되어 있으며, 단어 검증이나 필터링과 같은 단순한 작업에 사용됩니다.

WordNet: 단어 간의 관계와 어휘 정보를 포함한 복잡한 데이터베이스로, 의미 분석, 단어 유사도 계산 등 더 고차원적인 자연어 처리 작업에 사용됩니다.

nltk.ne_chunk(tags, binary=False)는 NLTK(Natural Language Toolkit) : 텍스트에서 **명명된 개체(named entities)**를 식별하고 태깅하는 작업을 수행합니다. 사람, 조직, 장소, 날짜 등과 같은 고유한 이름을 식별

pos_tag():

한계: 고유명사와 같은 특정한 명명된 개체를 인식하지만, 그 개체가 사람, 장소, 조직 등 어떤 종류인지에 대한 정보는 제공하지 않습니다.

# **Text 클래스**
- NLTK의 `Text` 클래스는 문서 분석에 유용한 여러가지 메서드를 제공한다. 토큰열을 입력하여 생성한다.

- Text() 함수는 NLTK(Natural Language Toolkit)에서 제공하는 클래스 중 하나로, 주어진 토큰 리스트를 래핑(wrapping)하여 텍스트 분석에 유용한 여러 메서드를 제공합니다. 구체적으로 Text()의 주요 기능은 다음과 같습니다:

**콘코던스(Concordance) 분석**:

특정 단어가 텍스트 내에서 어떻게 사용되는지 문맥을 확인할 수 있습니다.

예: text.concordance("Emma")는 "Emma"라는 단어가 등장하는 모든 문맥을 보여줍니다.

**단어의 빈도 분석**:

텍스트 내에서 단어들이 얼마나 자주 등장하는지 분석할 수 있습니다.

예: text.vocab()을 사용하여 단어 빈도 분포를 확인할 수 있습니다.

**유사 단어 찾기(Similar Words)**:

특정 단어와 비슷한 문맥에서 사용되는 단어들을 찾을 수 있습니다.
예: text.similar("happy")는 "happy"와 비슷한 맥락에서 사용되는 단어들을 보여줍니다.

**분산도(Dispersion Plot) 시각화**:

텍스트 내에서 특정 단어들의 분포를 시각적으로 표현할 수 있습니다.
예: text.dispersion_plot(["Emma", "Pemberley", "love"])는 지정된 단어들이 텍스트 내 어디에 위치하는지 그래프로 보여줍니다.

# **Bag of Words(BoW)를 이용한 벡터화**

```python
import re
from konlpy.tag import Mecab

corpus = '평생 살 것처럼 꿈을 꾸어라. 그리고, 내일 죽을 것처럼 오늘을 살아라!'
```

CountVectorizer는 텍스트 데이터를 수치 벡터화하여  Bag-of-Words(BOW) 모델을 구현하는 도구입니다.

텍스트 데이터의 각 단어의 빈도수(frequency)를 기반으로 특징을 생성

CountVectorizer는 다음과 같은 방식으로 작동합니다:

- 토큰화 (Tokenization):
입력된 텍스트를 개별 단어(토큰)로 분리합니다. 기본적으로 공백이나 구두점을 기준으로 단어를 분리합니다.

- 불용어 제거 (Stop Words Removal):
불용어(stop words)가 지정된 경우, 토큰화된 단어 목록에서 불용어를 제거합니다.

- 단어 빈도 계산 (Count Frequencies):
텍스트 데이터에서 각 단어의 출현 빈도를 계산합니다.

- 벡터 변환 (Vectorization):
계산된 단어 빈도를 기반으로 문서-단어 행렬(document-term matrix)을 생성합니다. 이 행렬의 각 행은 하나의 문서를 나타내고, 각 열은 특정 단어의 빈도를 나타냅니다.

어휘빈도-문서역빈도(TF-IDF) 분석

- 어휘 빈도-문서 역빈도(TF-IDF : Term Frequency-Inverse Document Frequency)는 단순히 빈도수가 높은 단어가 핵심어가 아닌, 특정 문서에서만 집중적으로 등장할 때 해당 단어가 문서의 주제를 잘 담고 있는 핵심어라고 가정
- 특정 문서에서 특정 단어가 많이 등장하고 그 단어가 다른 문서에서 적게 등장할 때, 그 단어를 특정 문서의 핵심어로 간주
- 어휘 빈도-문서 역빈도는 어휘빈도와 문서 역빈도를 곱해 계산 가능
- 어휘 빈도는 특정 단어가 문서 내에서 얼마나 자주 등장했는가를 나타낸다
- 문서 역빈도는 특정 단어가 전체 문서 집합에서 얼마나 자주 등장하는가를 반영
